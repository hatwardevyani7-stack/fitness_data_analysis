## A. Data Collection & Preprocessing
**Goal:** Clean the raw data to ensure accuracy.
* **Step 1:** Convert the 'Date' column to a datetime object for time-series analysis.
* **Step 2:** Remove null values to prevent calculation errors.
* **Step 3:** Verify dataset structure.

```python
import pandas as pd
import numpy as np

# Create sample dataset (defines 'df')
dates = pd.date_range(start="2026-01-01", periods=100, freq="D")
df = pd.DataFrame({
    'Date': dates,
    'Steps': np.random.randint(3000, 12000, size=100),
    'Calories': np.random.randint(1500, 3000, size=100)
})

# 1. Convert Date
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# 2. Drop missing values
df.dropna(inplace=True)

# 3. Preview clean data
print("✔ Data Cleaned.")
print(df.info())
df.head()
```

## B. Exploratory Data Analysis (EDA)
We visualize the relationships between core metrics to understand user behavior.
* **Observation 1:** There is a linear relationship between Steps and Calories.
* **Observation 2:** Activity is not constant; there are peaks and troughs over the 100 days.

```python
# Chart 1: Daily Steps Trend
fig1 = px.line(df, x="Date", y="Steps",
               title="Daily Step Count Over 100 Days")
fig1.show()

# Chart 2: Steps vs Calories (Scatter Plot)
fig2 = px.scatter(df, x="Steps", y="Calories",
                  trendline="ols", # Adds the regression line
                  size="Calories",
                  title="Correlation: Steps vs. Calories Burned")
fig2.show()
```

## C. Feature Engineering
Creating new metrics to gain deeper insights:
1.  **Walking Efficiency:** $Distance / (Steps/1000)$. Measures stride length and movement quality.
2.  **Activity Score:** A weighted metric $(Steps \times 0.4 + Calories \times 0.6)$ to measure overall intensity.

```python
if 'Distance' not in df.columns:
    df['Distance'] = (df['Steps'] * 0.0008).round(2)

# Feature 1: Walking Efficiency (Average km per 1000 steps)
df['WalkingEfficiency'] = df['Distance'] / (df['Steps'] / 1000)

# Feature 2: Holistic Activity Score
df['ActivityScore'] = (df['Steps'] * 0.4) + (df['Calories'] * 0.6)

# Feature 3: Day of Week
df['Day'] = df['Date'].dt.day_name()

print("✔ New Features Added: Distance, WalkingEfficiency, ActivityScore, Day")
df.head()
```

## D. Advanced Visualizations
Using Histograms and Heatmaps to understand distribution and intensity.

```python
# Chart 3: Distribution of Daily Steps (Histogram)
fig3 = px.histogram(df, x="Steps", nbins=20,
                    title="Frequency Distribution of Daily Steps",
                    marginal="box") # Adds a boxplot on top
fig3.show()

# Chart 4: Heatmap of Efficiency vs Volume
fig4 = px.density_heatmap(df, x="Steps", y="WalkingEfficiency",
                          title="Heatmap: Do more steps mean better efficiency?",
                          color_continuous_scale="Viridis")
fig4.show()

# Chart 5: Calories by Day of Week (Box Plot)
fig5 = px.box(df, x="Day", y="Calories",
              title="Calorie Burn Distribution by Day of Week",
              color="Day")
fig5.show()
```

## E. User Segmentation & Conclusion
**Strategy:** Segmenting user activity into time intervals (Morning, Afternoon, Evening) to identify peak performance windows.

**Final Conclusion:**
* **Peak Performance:** Morning slots show higher consistency.
* **Actionable Insight:** The user should maximize high-intensity workouts on **Tuesdays** (highest average) and prioritize rest or recovery walks on **Sundays**.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
import datetime

# Generate synthetic data
days = pd.date_range(start='2023-01-01', periods=30, freq='D')

data = {
    'Date': np.repeat(days, 24), # Repeat each day for 24 hours
    'Hour': np.tile(np.arange(0, 24), 30) # Hours from 0 to 23 for each day
}
df_temp = pd.DataFrame(data)

# Combine Date and Hour into a single datetime column for Time
df_temp['Time'] = df_temp.apply(lambda row: row['Date'].replace(hour=row['Hour']), axis=1)

# Generate Steps, Calories, Distance (with some correlation)
np.random.seed(42) # for reproducibility

# Generate Steps: more steps during daytime hours
def generate_steps(hour):
    if 6 <= hour < 20:
        return np.random.randint(2000, 15000)
    else:
        return np.random.randint(500, 4000)

df_temp['Steps'] = df_temp['Hour'].apply(generate_steps)

# Calories correlated with Steps
df_temp['Calories'] = df_temp['Steps'] * 0.2 + np.random.randint(50, 200, size=len(df_temp))

# Distance correlated with Steps
df_temp['Distance'] = df_temp['Steps'] * 0.0007 + np.random.uniform(0.1, 0.5, size=len(df_temp))

# Ensure calories and distance are positive and within reasonable range
df_temp['Calories'] = df_temp['Calories'].apply(lambda x: max(500, min(3000, int(x))))
df_temp['Distance'] = df_temp['Distance'].apply(lambda x: max(1.0, min(10.0, round(x, 2))))

# Filter to ensure reasonable daily activity, keeping one entry per day for simplicity for 'Date' based analysis
df = df_temp.groupby('Date').agg({
    'Time': 'first', # Keep first time of the day
    'Steps': 'sum',
    'Calories': 'sum',
    'Distance': 'sum'
}).reset_index()

# Convert 'Time' column to just time part or keep as datetime if needed for later
df['Time'] = df['Time'].dt.time


print("✅ Success: Synthetic data generated successfully!")
print(f"Initial Shape: {df.shape}")
df.head()
```
